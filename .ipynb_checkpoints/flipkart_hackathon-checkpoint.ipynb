## Dataset Preprocessing

#### Removing NaN values

```python
df_train.isna().sum()
```

```python
numerical_cols = df_train.select_dtypes(include = ['int64','float64']).columns.tolist()
```

```python
categorical_cols = df_train.select_dtypes(include = ['object']).columns.tolist()
```

```python
print("Numerical:", numerical_cols)
print("Categorical:", categorical_cols)
```

```python
for col in numerical_cols:
    df_train[col] = df_train[col].fillna(df_train[col].median())

for col in categorical_cols:
    mode = df_train[col].mode()[0]
    df_train[col] = df_train[col].fillna(mode)
```

```python
df_train.isna().sum()
```

#### Encoding

```python
categorical_cols
```

```python
for col in categorical_cols:
    print(col, ":", df_train[col].nunique())
```

```python
le = LabelEncoder()
le.fit(df_train['geohash'])
df_train['geohash'] = le.transform(df_train['geohash'])
```

```python
df_train.head()
```

```python
df_train['timestamp'].unique()
```

```python
ohe_cols = ['RoadType','Weather']

df_train = pd.get_dummies(
    df_train,
    columns=ohe_cols,
    dtype=int,
    drop_first=True
)
df_train.head()
```

```python
df_train['LargeVehicles'] = df_train['LargeVehicles'].map({
    'Not Allowed': 0,
    'Allowed': 1
})

df_train['Landmarks'] = df_train['Landmarks'].map({
    'No': 0,
    'Yes': 1
})
```

```python
df_train.head()
```

```python
df_train.columns
```

#### Convert time stamp to individual features hour and minutes

```python
split = df_train['timestamp'].str.split(':',expand=True)
df_train['hour'] = split[0].astype(int)
df_train['minute'] = split[1].astype(int)

df_train.drop('timestamp', axis=1, inplace=True)
```
